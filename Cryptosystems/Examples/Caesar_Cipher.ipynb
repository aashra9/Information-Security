# Caesar Cipher 

This notebook implements encryption and decryption using the Caesar cipher.

- **Encryption:** shifts each letter forward by the key.
- **Decryption:** shifts each letter backward by the key.
- Uppercase/lowercase letters are preserved.
- Non-alphabetic characters are unchanged.

```python
# Things to note

The built-in ord() function takes a single Unicode character and returns its corresponding Unicode code point value (which matches the standard ASCII value for basic English characters).Inverse Function: The opposite of ord() is chr(). Running chr(65) will return the string 'A'.Error Trigger: You must pass exactly one character. Passing an empty string or multiple characters like ord('AB') will throw a TypeError
```

```python
def caesar_encrypt(text, key):
    result = ""

    for char in text:
        if char.isupper():
            result += chr((ord(char) - ord('A') + key) % 26 + ord('A'))
        elif char.islower():
            result += chr((ord(char) - ord('a') + key) % 26 + ord('a'))
        else:
            result += char

    return result


def caesar_decrypt(ciphertext, key):
    return caesar_encrypt(ciphertext, -key)


message = "Hello, World!"
key = 3

ciphertext = caesar_encrypt(message, key)
plaintext = caesar_decrypt(ciphertext, key)

print("Plaintext :", message)
print("Key       :", key)
print("Ciphertext:", ciphertext)
print("Decrypted :", plaintext)
```
