```python
def sieve(n):
    """Return all primes <= n."""
    is_prime = [True] * (n + 1)
    is_prime[0] = is_prime[1] = False

    p = 2
    while p * p <= n:
        if is_prime[p]:
            for k in range(p * p, n + 1, p):
                is_prime[k] = False
        p += 1

    return [n for n in range(2, n + 1) if is_prime[n]]
```

```python
primes = sieve(100)

print(primes)
print(len(primes))
```

```python
import bisect

primes = sieve(1_000_000)

def pi(x):
    return bisect.bisect_right(primes, x)
```

```python
for x in [10, 100, 1_000, 10_000, 100_000, 1_000_000]:
    print(x, pi(x))
```

```python
import matplotlib.pyplot as plt
import numpy as np

xs = np.logspace(1, 6, 500)

actual = np.array([pi(int(x)) for x in xs])
pnt = xs / np.log(xs)

plt.figure(figsize=(10, 6))

plt.plot(xs, actual, label=r"$\pi(x)$")
plt.plot(xs, pnt, label=r"$x/\log x$")

plt.xscale("log")
plt.xlabel("x")
plt.ylabel("Number of primes")
plt.title("Prime Number Theorem")
plt.legend()
plt.grid(True)

plt.show()
```
