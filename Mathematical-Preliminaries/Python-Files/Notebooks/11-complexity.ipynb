Jupyter notebook demonstrating big-O growth, complexity classes, and concrete security parameter estimates.

## 1. Big-O growth rates

Compare the growth of common functions for increasing input sizes.

```python
import math

print(f"{'n':>10} {'log2 n':>10} {'n^2':>12} {'2^n':>12}")
for n in [10, 30, 60, 100]:
    print(f"{n:>10} {math.log2(n):>10.2f} {n*n:>12} {2**n:>12.2e}")
```

## 2. Enumeration of subsets vs permutations

Verify that brute-forcing search spaces quickly exceeds computational limits.

```python
import math

for n in [10, 20, 40, 64, 128]:
    subsets = 2**n
    print(f"n={n}: 2^n = {subsets:.2e} subsets")
```

## 3. Concrete security / work factor

Estimate how long it takes to brute force a keyspace given hashes per second.

```python
work = 3e9 * 3600 * 24 * 365  # 3 billion hashes/sec for a year
for bits in [40, 56, 64, 80, 128, 256]:
    keyspace = 2**bits
    years = keyspace / work
    print(f"{bits}-bit key: keyspace 2^{bits} ~ {keyspace:.1e}, brute force ~ {years:.1e} years")
```

## 4. Asymmetric vs symmetric key sizes (NIST equivalence)

Ballpark equivalent security levels for RSA/ECC vs symmetric.

```python
for sym, rsa, ecc in [(80, 1024, 160), (112, 2048, 224), (128, 3072, 256), (256, 15360, 512)]:
    print(f"Symmetric {sym} bits ~ RSA {rsa} bits ~ ECC {ecc} bits")
```
