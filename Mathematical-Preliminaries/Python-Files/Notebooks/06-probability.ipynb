## 1. Union bound (Boole's inequality)

The probability of a union of events is at most the sum of the individual probabilities:  
$P(\bigcup_i E_i) \le \sum_i P(E_i)$.

```python
import itertools, random

# Estimate P(A or B) for two independent events with p=0.3 each
p = 0.3
N = 200000
hits = 0
for _ in range(N):
    A = random.random() < p
    B = random.random() < p
    if A or B:
        hits += 1
empirical = hits / N
exact = 1 - (1-p)**2
print(f"P(A U B) empirical  = {empirical:.4f}")
print(f"P(A U B) exact      = {exact:.4f}")
print(f"Union bound sum     = {2*p:.4f}  (>= exact, as expected)")
```

## 2. Conditional probability and Bayes' theorem

Given an imperfect test, compute $P(\text{has disease} \mid \text{positive})$.

```python
# Prevalence P(D) = 0.01, sensitivity P(T+|D) = 0.99, specificity P(T-|!D) = 0.95
P_D = 0.01
P_Tpos_given_D = 0.99
P_Tpos_given_notD = 1 - 0.95

# Bayes: P(D|T+) = P(T+|D) P(D) / P(T+)
P_Tpos = P_Tpos_given_D * P_D + P_Tpos_given_notD * (1 - P_D)
P_D_given_Tpos = P_Tpos_given_D * P_D / P_Tpos
print(f"P(T+) = {P_Tpos:.4f}")
print(f"P(D | T+) = {P_D_given_Tpos:.4f}")
```

## 3. Random variables and expectation (linearity)

For a random permutation, compute the expected number of fixed points, which is 1 by linearity of expectation.

```python
import statistics

def fixed_points(perm):
    return sum(1 for i, x in enumerate(perm) if x == i)

n = 10
N = 50000
counts = []
for _ in range(N):
    perm = list(range(n))
    random.shuffle(perm)
    counts.append(fixed_points(perm))

print(f"Empirical mean fixed points: {statistics.mean(counts):.4f}  (theoretically 1)")
```

## 4. Independence vs. correlation

Two independent random variables have zero covariance. Verify empirically.

```python
Xs = [random.random() for _ in range(N)]
Ys = [random.random() for _ in range(N)]

mean_x = sum(Xs) / N
mean_y = sum(Ys) / N
cov = sum((x-mean_x)*(y-mean_y) for x, y in zip(Xs, Ys)) / N
print(f"Empirical covariance of two independent vars: {cov:.4f} (near 0)")
```

## 5. Chernoff / random sampling

Estimate a population proportion by random sampling and compare with the theoretical confidence interval.

```python
import math

true_p = 0.5
SAMPLE = 10000
samples = [1 if random.random() < true_p else 0 for _ in range(SAMPLE)]
est = sum(samples) / SAMPLE
# 95% confidence: est ± 1.96*sqrt(p(1-p)/n)
se = math.sqrt(est*(1-est)/SAMPLE)
print(f"Estimated p = {est:.4f}, margin = {1.96*se:.4f}")
print(f"95% CI = [{est-1.96*se:.4f}, {est+1.96*se:.4f}], true = {true_p}")
```
